```python
# whole module is imported
import math

print(math.sqrt(25))
print(math.factorial(5))
```

```python
# specific module imported
from math import sqrt, factorial

print(sqrt(36))
print(factorial(4))

```

```python
# import all functions from package

from math import *

print(sqrt(64))
```

```python
# import package with alias

import math as m

print(m.factorial(10))
print(m.sin(90))
```

```python
# random password geneartor

import random
import string


letters=string.ascii_letters
digits=string.digits
symbols="!@#$%"

all_chars=letters+digits+symbols

password=""

for i in range(10):
    password+=random.choice(all_chars)

print("Generated PAssword is :: ",password)
```

```python
# dice game simulator
 
import random

dice1=random.randint(1,6)
dice2=random.randint(1,6)

print("Dice 1 :",dice1)
print("Dice 2 :",dice2)

print("Total is :: ",dice1+dice2)

```

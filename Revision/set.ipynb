```python
# manual union

a={10,20,30}
b={30,40,50}

union=set()

for i in a:
    union.add(i)

for i in b:
    if i not in union:
        union.add(i)


print(union)   
```

```python
# intersection

a={10,20,30,40}
b={30,40,50,60}

res=set()

for i in a:
    if i in b:
        res.add(i)

print(res)
```

```python
#difference a-b 

a={10,20,30}
b={20,30,40}

res=set()

for i in a:
    if i not in b:
        res.add(i)

print(res)
```

```python
# remove duplicates from list 
a=[10,20,20,10,10,10,20,30,40,50]

res=set()
for i in a:
    if i not in res:
        res.add(i)

print(res)
```

```python
# two set are disjoint -- unique 


a={10,20,3}
b={40,50,60}

is_disjoint=True

for i in a:
    if i in b:
        is_disjoint=False
        break

print("Disjoint Sets" if is_disjoint else "Not a DisJoint Set ")
```

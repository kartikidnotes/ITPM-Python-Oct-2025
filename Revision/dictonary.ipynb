```python
#count the frequency 

s="aabbbbbccbbb"
res={}

for i in s:
    if i in res:
        res[i]+=1
    else:
        res[i]=1

print(res)
```

```python
# highest key value

d={"a":10,"b":90,"c":30}

max_key=""
max_val=0

for k in d:
    if d[k]>max_val:
        max_val=d[k]
        max_key=k

print(max_key)
print(max_val)
```

```python
# Merge 

d1={"a":1,"b":2,"c":3}
d2={"name":"raj","course":"mern","b":20}

result={}

for i in d1:
    result[i]=d1[i]

for i in d2:
    result[i]=d2[i]

print(result)
```

```python
#swap key value pair

d={"id":"1","course":"mern","marks":70}

swap={}

for i in d:
    value=d[i]
    swap[value]=i

print(swap)
```

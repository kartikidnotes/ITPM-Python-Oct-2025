```python
# Reverse the string 

s="Python Programming"
rev=""

for ch in s:
    rev=ch+rev

print(rev)
```

```python
# count the vowels

s="India is my country"

count=0
for ch in s:
    if ch in "aeiouAEIOU":
        count+=1

print(count)
```

```python
# remove the duplicates
s="india is my country"
res=""

for ch in s:
    if ch not in res:
        res+=ch


print(res)
```

```python
# check palindrome

s="dad"
rev=""

for ch in s:
    rev=ch+rev

print("String is Palindrome " if s==rev else "NOT a Palindrome String ")
```

```python
# most frequent character 

s="aabbbbccbbbbc"
max_count=0
max_char=""

for ch in s:
    count=0
    for c in s:
        if ch==c:
            count+=1
    if count>max_count:
        max_count=count
        max_char=ch

print(max_char)
```

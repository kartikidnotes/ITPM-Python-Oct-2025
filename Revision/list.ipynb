```python
# find second highest element

lst=[10,20,5,8]

largest=second=-9999

for num in lst:
    if num>largest:
        second=largest
        largest=num
    elif num>second and num!=largest:
        second=num


print(second)
```

```python
# remove the duplicates

lst=[10,20,20,20,30,30,40,40,50]

res=[]

for i in lst:
    if i not in res:
        res.append(i)


print(res)
```

```python
# common elements from list

a=[10,20,30]
b=[20,30,40]

common=[]

for i in a:
    if i in b:
        common.append(i)


print(common)

```

```python
# move all 0 to end

lst=[0,10,0,20,0,30,12]
res=[]

for i in lst:
    if i!=0:
        res.append(i)

for i in lst:
    if i==0:
        res.append(i)

print(res)
```

```python
# find duplicate

lst=[10,20,30,30,30,40,50,60,60]
dup=[]

for i in lst:
    if lst.count(i)>1 and i not in dup:
        dup.append(i)

print(dup)
```

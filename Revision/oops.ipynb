```python
# parameterized constructor

class Student:
    def __init__(self,name,course,marks):
        self.name=name
        self.course=course
        self.marks=marks
        
    def show(self):
        print("Name is :: ",self.name)
        print("Course is :: ",self.course)
        print("Marks is :: ",self.marks)

s1=Student("Ram","MERN",99)
s2=Student("Raj","MEAN",90)

s1.show()
s2.show()
```

```python
# abstraction

from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def area(self):
        side=5
        print("Area of Square is :: ",side*side)

class Rectangle(Shape):
    def area(self):
        l=10
        b=20
        print("Area of Rectangle is :: ",l*b)

s=Square()
s.area()

r=Rectangle()
r.area()
```

```python
from abc import ABC, abstractmethod

class Money(ABC):
    @abstractmethod
    def withdraw(self):
        pass


class Bank(Money):
    def withdraw(self):
        print("Money is Withdrawn with help of withdraw slip ")

class ATM(Money):
    def withdraw(self):
        print("Moeny is withdrawn from ATM Machine ")

b=Bank()
a=ATM()

b.withdraw()
a.withdraw()
```

```python
# Single Inheritance 

class Parent():
    def show(self):
        print("I'm from Parent Class")

class Child(Parent):
    def display(self):
        print("I'm from Child CLAsss")

c=Child()
c.show()
c.display()
```

```python
# multilevel

class Grandparent:
    def gp(self):
        print("Grandparent Class is Called ")

class Parent(Grandparent):
    def p(self):
        print("Parent Class is called ")

class Child(Parent):
    def c(self):
        print("Child CLass is called ")

obj=Child()
obj.c()
obj.gp()
obj.p()
```

```python
# multiple

class Father:
    def fatherskill(self):
        print("Driving very Well ")

class Mother:
    def motherskill(self):
        print("Cooking Very welll")

class Child(Father,Mother):
    def childskill(self):
        print("study very well ")

obj=Child()
obj.fatherskill()
obj.motherskill()
obj.childskill()
```

```python
# hierarchical 

class Parent:
    def show(self):
        print("In Parent class")

class Child1(Parent):
    def c1(self):
        print("In CHILD 1 Class ")

class Child2(Parent):
    def c2(self):
        print("In CHILD 2 Class ")

ob1=Child1()
ob2=Child2()

ob1.show()
ob1.c1()

ob2.show()
ob2.c2()
```

```python
# hybrid inheritance

class A:
    def show(self):
        print("In A CLass")

class B(A):
    def bshow(self):
        print("In class B ")

class C(A):
    def cshow(self):
        print("In class C ")

class D(B,C):
    def dshow(self):
        print("In class D")

d=D()
d.show()
d.bshow()
d.cshow()
d.dshow()
```

```python
# method overloading

class Operation:
    def add(self,a=0,b=0,c=0):
        print(a+b+c)

obj=Operation()
obj.add(2,3)
obj.add(7)
obj.add(1,2,3)
```

```python
# overriding

class Parent:
    def show(self):
        print("In Parent Class ")

class Child(Parent):
    def show(self):
        print("In Child CLass ")

obj=Child()
obj.show()
```

```python
# duck type

class CreditCard:
    def pay(self,amt):
        print(f"Rs. {amt} paid by Credit Card ")

class UPI:
    def pay(self,amt):
        print(f"Rs. {amt} paid by UPI ")
    
class Cash:
    def pay(self,amt):
        print(f"Rs. {amt} paid by Cash ")

def  make_payment(method,amount):
    method.pay(amount)

make_payment(CreditCard(),50000)
make_payment(UPI(),10000)
make_payment(Cash(),5000)
```

```python
# Multiple Exception

try:
    a=int(input("Enter number :: "))
    b=int(input("Enter number :: "))
    print(a/b)
except ZeroDivisionError:
    print("Cannot divide a number by 0")
except ValueError:
    print("Invalid Input")
```

```python
# else

try:
    a=int(input("Enter number :: "))
except ValueError:
    print("Invalid")
else:
    print("Number is :: ",a)
   
```

```python
# without argument

def demo():
    print("Hello this is a demo function")

demo()
```

```python
# with argument

def add(a,b):
    print(a+b)


add(2,3)


def square(side):
    print(side*side)

square(5)
```

```python
# func with argument and return type

def multiply(a,b,c):
    return a*b*c


res=multiply(2,3,4)
print(res)
```

```python
# func with return but with no argument

def greet():
    return "Hello From Greet Function "

print(greet())
```

```python
# default argument

def demo(name="Ram"):
    print("Hello ",name)


demo()
demo("Kartiki")
```

```python
# keyword Arguments

def student(name,marks):
    print(name,marks)


student(name="Ram",marks=99)
student(marks=90,name="Raj")
```

```python
# varibale length
def total(*numbers):
    sum=0
    for i in numbers:
        sum+=i
    print(sum)


total(10,20)
total(10,20,30)
total(10,20,30,40,50,60,70,80)
```

```python
# keyword arguments

def stud_details(**data):
    for i in data:
        print(i," : ",data[i])

stud_details(name="Ram",city="Pune",marks=99)
```

```python
#lamba function

square=lambda a:a*a
print(square(5))
```

```python
# Recursion

def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)


print(factorial(5))
```

```python
# nested funtion
def outer():
    def add():
        print("Inside the add function ")
    add()

outer() 
```

```python
# higher order function

def double(num):
    return num*2

def apply(func,value):
    return func(value)

print(apply(double,5))
```

```python
# map

lst=[1,2,3,4,5,6]
res=list(map(lambda x:x*3,lst))

print(res)
```

```python
#filter
lst=[1,2,3,4,5,6]

res=list(filter(lambda x:x%2==0,lst))

print(res)
```

```python
# reduce
from functools import reduce
lst=[1,2,3,4,5,6]

res=reduce(lambda x,y: x+y,lst)

print(res)
```

```python
# convert tuple to list -- update 
t=(10,20,30)
lst=[]

for i in t:
    lst.append(i)

lst.append(40)
t=tuple(lst)
print(t)
```

```python
# count element 

t=(10,20,30,40,50)
count=0

for _ in t:
    count+=1

print(count)
```

```python
#max element
t=(10,50,90,40)
max_val=t[0]

for i in t:
    if i>max_val:
        max_val=i

print(max_val)
```

```python
# removing tuuple duplicate

t=(10,20,20,30,40,30,40,50)

res=()

for i in t:
    if i not in res:
        res=res+(i,)

print(res)
```

```python
# frequency of elements 
t=(10,20,20,30,30,30,40,40,40,40,40,40,60,50,50,10)

freq={}

for i in t:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1

print(freq)
```
